```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def show_comparision(original,blur3,blur5,blur7):
  plt.figure(figsize=(12,10))
  #original img
  plt.subplot(2,2,1)
  plt.imshow(original,cmap="gray")
  plt.title('Original Grayscale image')
  plt.axis('off')
  #filtered image 3*3 kernal
  plt.subplot(2,2,2)
  plt.imshow(blur3,cmap="gray")
  plt.title('3x3 Filter')
  plt.axis('off')
  #5*5 kernal
  plt.subplot(2,2,3)
  plt.imshow(blur5,cmap="gray")
  plt.title('5x5 Filter')
  plt.axis('off')
  #7*7 kernal
  plt.subplot(2,2,4)
  plt.imshow(blur7,cmap="gray")
  plt.title('7x7 Filter')
  plt.axis('off')
  plt.show()
```

```python
img=cv2.imread('/content/images.jpg')
```

```python
kernel_3_3=np.ones((3,3),np.float32)/9
kernel_5_5=np.ones((5,5),np.float32)/25
kernel_7_7=np.ones((7,7),np.float32)/49
```

```python
if img is None:
    print("Error: The image could not be loaded. Please check if '/content/effel.jpg' exists and the path is correct.")
else:
    blur_3_3=cv2.filter2D(img,-1,kernel_3_3)
    blur_5_5=cv2.filter2D(img,-1,kernel_5_5)
    blur_7_7=cv2.filter2D(img,-1,kernel_7_7)
```

```python
show_comparision(img,blur_3_3,blur_5_5,blur_7_7)
```
